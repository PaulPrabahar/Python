QUESTION 1: TUPLE DETAILS

Create a tuple containing the names of five cities.
Perform the following operations:
- Print the tuple.
- Display the city at index 2.
- Display the last three cities using slicing.
- Check how many times a selected city appears using count().
- Find the index of a selected city using index().

```python
cities_tuple = ("chennai", "munbai", "bengaluru", "hydrabad", "kochi")

print(cities_tuple)
print(cities_tuple[2])
print(cities_tuple[2:])
print(cities_tuple.count("chennai"))
print(cities_tuple.index("kochi"))
```

QUESTION 2: TUPLE UPDATE THROUGH LIST CONVERSION

Create a tuple containing the values 10, 20, 30, 40, and 50.
Since tuples are immutable:
- Convert the tuple into a list.
- Replace 30 with 35.
- Add 60 to the list.
- Remove the first element.
- Convert the updated list back into a tuple.
- Print the final tuple.

```python
number_tuple = (10, 20, 30, 40, 50)
converted_tuple = list(number_tuple)
converted_tuple[2] = 35 
converted_tuple.append(60)
converted_tuple.remove(10)
converted_tuple = tuple(converted_tuple)
print(type(converted_tuple))
print(converted_tuple)
```

QUESTION 3: COMBINE AND PROCESS TUPLES

Create two tuples:
colors1 = ("red", "blue", "green")
colors2 = ("yellow", "blue", "white")

Perform the following operations:
- Combine both tuples into a new tuple.
- Print the combined tuple.
- Count how many times "blue" appears.
- Display the combined tuple using slicing to show alternate elements.

```python
colors1 = ("red", "blue", "green")
colors2 = ("yellow", "blue", "white")

total_color = colors1 + colors2
print(total_color)
print(f"total times the color blue appeared {total_color.count("blue")}")
print(total_color[::2])
```

QUESTION 4: FIND COMMON AND UNIQUE ELEMENTS

Create two sets:
morning_batch = {"Arun", "Bala", "Charan", "Divya"}
evening_batch = {"Bala", "Divya", "Ezhil", "Farah"}

Perform the following operations:
- Display students present in either batch.
- Display students present in both batches.
- Display students present only in the morning batch.
- Display students present only in the evening batch.

```python
morning_batch = {"Arun", "Bala", "Charan", "Divya"}
evening_batch = {"Bala", "Divya", "Ezhil", "Farah"}

print(morning_batch.union(evening_batch))
print(morning_batch.intersection(evening_batch))
print(morning_batch.difference(evening_batch))
print(evening_batch.difference(morning_batch))
```

QUESTION 6: UPDATE PRODUCT SETS

Create two sets:
available_products = {"Laptop", "Mouse", "Keyboard"}
new_products = {"Monitor", "Mouse", "Printer"}

Perform the following operations:
- Add "Webcam" to available_products.
- Remove "Keyboard" from available_products.
- Combine available_products and new_products using the union operator.
- Find the products common to both sets using the intersection operator.
- Find the products available only in available_products using the difference operator.

```python
available_products = {"Laptop", "Mouse", "Keyboard"}
new_products = {"Monitor", "Mouse", "Printer"}

available_products.add("Webcam")
print(available_products)
available_products.remove("Keyboard")
print(available_products)

combined_products = available_products.union(new_products)
common_products = available_products.intersection(new_products)
products_only_in_available = available_products.difference(new_products)
print(combined_products)
print(common_products)
print(products_only_in_available)
```

QUESTION 9: MANAGE A SHOPPING CART DICTIONARY

Create a dictionary for a shopping cart:

cart = {"Pen": 10, "Notebook": 50, "Bag": 500}

Perform the following operations:
- Add "Pencil" with a value of 5.
- Update the value of "Notebook" to 60 using update().
- Retrieve the value of "Bag" using get().
- Remove "Pen" using pop() and display the removed value.
- Display the remaining dictionary.
- Use clear() to remove all items and print the empty dictionary.

```python
cart = {"Pen": 10, "Notebook": 50, "Bag": 500}
cart.setdefault("Pencil", 5)
cart.update({"Notebook": 60})
print(cart.get("Bag"))
cart.pop("Pen")
print(cart)
cart.clear()
print(cart)
```

QUESTION 10: CREATE A SUBJECT MARKS DICTIONARY USING ZIP

marks = [90, 85, 95, 88]

- Create a dictionary using dict(zip(subjects, marks)).
