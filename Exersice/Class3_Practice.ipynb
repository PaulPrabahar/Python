### Problem 1: Voting eligibility

A person can vote only if they are 18 or older **and** are a citizen. Given `age = 25` and `citizen = True`, print `Eligible to vote`. If the age is below 18 print `Under age`. If 18 or older but not a citizen print `Not a citizen`. Use a nested `if`. Then test with `citizen = False` and `age = 16`.

```python
age = 25
citizen = True

if age >= 18:
    if citizen:
        print("Eligible to vote")
    else:
        print("Not a citizen")
else:
    print("Under age")
```

### Problem 2: Simple calculator

Given `a = 12`, `b = 4` and `operator = "/"`, print the result for `+`, `-`, `*` and `/`. For any other operator print `Invalid operator`. Test with each operator.

```python
a = 12
b = 4
operator = "+"

if operator == "+":
    print(a + b)
elif operator == "-":
    print(a - b)
elif operator == "*":
    print(a * b)
elif operator == "/":
    print(a / b)
else:
    print("Invalid operator")
```

### Problem 3: Divisible by 3 and 7

Given `num = 42`, print `Divisible by 3 and 7` if the number is divisible by both, otherwise print `Not divisible`. Test with `21`, `14` and `42`.

```python
num = 42

if num % 3 == 0 and num % 7 == 0:
    print("Divisible by 3 and 7")
else:
    print("Not divisible")
```

### Problem 4: Temperature category

Given `temp = 28` (in Celsius), print `Cold` if below 15, `Warm` if from 15 to 30 (both included), and `Hot` if above 30.

```python
temp = 28

if temp < 15:
    print("Cold")
elif temp <= 30:
    print("Warm")
else:
    print("Hot")
```

### Problem 5: Traffic signal

Given `signal = "yellow"`, print `Stop` for `red`, `Get ready` for `yellow`, `Go` for `green`, and `Invalid signal` for anything else. Make it work even if the value is typed as `Red` or `GREEN` (hint: a string method from Class 2).

```python
signal = "yellow"

if signal.lower() == "red":
    print("Stop")
elif signal.lower() == "yellow":
    print("Get ready")
elif signal.lower() == "green":
    print("Go")
else:
    print("Invalid signal")
```

### Problem 6: ATM withdrawal

Given `balance = 5000` and `withdraw = 2000`: if `withdraw` is more than 0 and not more than the balance, print `Withdraw successful` and the remaining balance. If it is more than the balance print `Insufficient balance`. If it is 0 or negative print `Invalid amount`. Test with `6000` and `-100`.

```python
balance = 5000
withdraw = 2000

if withdraw > 0:
    if withdraw <= balance:
        print("Withdraw successful")
        print("Remaining balance:", balance - withdraw)
    else:
        print("Insufficient balance")
else:
    print("Invalid amount")
```

### Problem 7: Marriage eligibility

The legal age is 21 for males and 18 for females. Given `gender = "female"` and `age = 19`, print `Eligible for marriage` or `Not eligible`. Test with `gender = "male"` and `age = 19`.

```python
gender = "female"
age = 19

if gender == "male" and age >= 21:
    print("Eligible for marriage")
elif gender == "female" and age >= 18:
    print("Eligible for marriage")
else:
    print("Not eligible")
```

**Note:** A shorter version joins both cases with `or`: `(gender == "male" and age >= 21) or (gender == "female" and age >= 18)`.

### Problem 8: Smallest of three numbers

Given `a = 8`, `b = 3`, `c = 15`, print the smallest number. Do not use `min()`. Test with `a = 2, b = 9, c = 5` and `a = 7, b = 7, c = 4`.

### Problem 9: Electricity bill

The rate depends on the units used: up to 100 units costs 2 per unit, up to 200 units costs 3 per unit, and above 200 units costs 5 per unit (the rate applies to all units). Given `units = 150`, print the bill like `Bill amount: 450`. Test with `80` and `250`.

```python
units = 150

if units <= 100:
    bill = units * 2
elif units <= 200:
    bill = units * 3
else:
    bill = units * 5

print("Bill amount:", bill)
```

### Problem 10: Type of the value

Given `x = 7.5`, print `Integer` if `x` is an int, `Float` if it is a float, `String` if it is a str, and `Boolean` if it is a bool. Test with `10`, `"hello"` and `True`.

```python
x = 7.5

if type(x) == int:
    print("Integer")
elif type(x) == float:
    print("Float")
elif type(x) == str:
    print("String")
elif type(x) == bool:
    print("Boolean")
else:
    print("Other type")
```
