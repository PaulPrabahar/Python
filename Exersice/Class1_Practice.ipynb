## B. Data types and Type casting

### Problem 6: String + string vs number + number

What do `"70" + "80"` and `70 + 80` print? Predict first, then run both. Then fix the first one so that it prints `150`.

````python
print("70" + "80")          
print(70 + 80)              
print(int("70") + int("80"))  
````

**Note:** `+` joins strings but adds numbers. Cast with `int()` before doing maths on text.

### Problem 7: Cast the marks

`mark1 = "70.10"` and `mark2 = "80.10"` are strings. Print (a) their sum as floats, (b) their sum as ints (convert via float first, then int).

````python
mark1 = "70.10"
mark2 = "80.10"

print(float(mark1) + float(mark2))      # 150.2
print(int(float(mark1)) + int(float(mark2)))  # 70 + 80 = 150
````

### Problem 8: Implicit casting

Without running: what is the type of `10 + 2.5`? And `True + True`? Then verify with `type()`.

**Solution:**

````python
print(10 + 2.5, type(10 + 2.5))        # 12.5 <class 'float'>  (int promoted to float)
print(True + True, type(True + True))  # 2 <class 'int'>       (True == 1)
````

## C. print() formatting and f-strings

### Problem 11: Different ways to print

With `name = "Paul"` and `age = 25`, print `Hello Paul, I'm 25 years old` in three ways: (a) commas in print, (b) `+` concatenation, (c) f-string.

**Solution:**

````python
name = "Paul"
age = 25

print(f"Hello {name}, I'm {age} years old")           # (c) f-string
````

**Note:** In (b) you must convert `age` with `str()`, otherwise `str + int` raises `TypeError`.

### Problem 12: Expressions inside f-strings

With `name = "Paul"`, `mark1 = 70`, `mark2 = 80.5`, print: `paul scored a total of 150.5 (average 75.25)`. Do the maths inside the f-string.

**Solution:**

````python
name = "paul"
mark1 = 70
mark2 = 80.5

print(f"{name.upper()} scored a total of {mark1 + mark2} (average {(mark1 + mark2) / 2})")
````

### Problem 13: Multi-line and escape characters

Print exactly this using ONE print call with `\n`, then again with a triple-quoted f-string:
```
Name: Paul
Age: 25
City: Chennai
```

**Solution:**

````python
name, age, city = "paul", 25, "Chennai"

print(f"Name: {name}\nAge: {age}\nCity: {city}")
print("------------------------------")

print(f"""Name: {name}
Age: {age}
City: {city}""")


print("------------------------------")
print("Hello\tworld")
````

### Problem 14: Quotes inside strings

Print `I'm 36 years old` and `He said "hi"` using the right quote style (no backslashes needed).

````python
print("I'm 36 years old")      # double quotes outside, apostrophe inside
print('He said "hi"')          # single quotes outside, double inside
print("He said \"hi\"")  
````

### Problem 15: Formatting numbers

Given `price = 1234.5678`, print it rounded to 2 decimals, and with thousand separators (e.g. `1,234.57`). Hint: use format specs `:.2f` and `:,.2f`.

````python
price = 1234.5678
print(f"{price:.2f}")     # 1234.57
print(f"{price:,.2f}")    # 1,234.57
````
