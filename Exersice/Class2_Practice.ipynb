## D. String methods

### Problem 1: Case methods

For `s = " pythoN is Good "` print: upper, lower, title, capitalize, swapcase. Notice what `capitalize()` does with the leading space.

**Solution:**

````python
s = " pythoN is Good "
print(repr(s.upper()))
print(repr(s.lower()))
print(repr(s.title()))
print(repr(s.capitalize()))  
print(repr(s.swapcase()))
````

### Problem 2: Strip variants

Clean `"   hello   "` three ways: remove both sides, only left, only right. Print each with `repr()` so spaces are visible.

**Solution:**

````python
s = "   hello   "
print(repr(s.strip()))
print(repr(s.lstrip()))
print(repr(s.rstrip()))
````

### Problem 27: Time converter

Given `seconds = 98765`, print it as `D days, H hours, M minutes, S seconds`.

**Solution:**

````python
seconds = 98765
days = seconds // 86400
rem = seconds % 86400
hours = rem // 3600
rem = rem % 3600
minutes = rem // 60
secs = rem % 60
print(f"{days} days, {hours} hours, {minutes} minutes, {secs} seconds")
````
